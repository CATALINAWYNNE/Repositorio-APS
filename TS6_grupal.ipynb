### Introducción
El presente trabajo tiene como objetivo identificar mediante el análisis teórico frente a que tipo de filtro nos encontramos.

A partir de los parámetros de diseño especificados ($\omega_0 = 1 \text{ rad/s}$, $R = 1\ \Omega$ y factor de calidad $Q = 2\sqrt{2}$), se realiza el estudio analítico en el dominio de la frecuencia compleja $s$ utilizando la transformada de Laplace.

Un filtro  es un circuito diseñado para seleccionar o modificar componentes espectrales de una señal en función de su frecuencia. En redes pasivas de segundo orden ($RLC$), el comportamiento dinámico se determina analizando la posición de los polos y ceros de la función de transferencia $H(s)$ en el plano complejo $s$.

### 1. Filtro Pasa-Bajos

Permite el paso de señales con frecuencias inferiores a la frecuencia de corte ($\omega < \omega_c$) y atenúa las componentes de frecuencia superior.
* **Forma Canónica ($2^\circ$ orden):**
  $$H(s) = \frac{\omega_0^2}{s^2 + \frac{\omega_0}{Q}s + \omega_0^2}$$

### 2. Filtro Pasa-Altos

 Atenúa las señales con frecuencias inferiores a la frecuencia de corte ($\omega < \omega_c$) y permite el paso libre de las frecuencias superiores.
* **Forma Canónica ($2^\circ$ orden):**
  $$H(s) = \frac{s^2}{s^2 + \frac{\omega_0}{Q}s + \omega_0^2}$$


### 3. Filtro Pasa-Banda

Permite el paso de un rango delimitado de frecuencias comprendido entre una frecuencia de corte inferior ($\omega_{c1}$) y una superior ($\omega_{c2}$), atenuando tanto frecuencias muy bajas como muy altas.
* **Forma Canónica ($2^\circ$ orden):**
  $$H(s) = \frac{\frac{\omega_0}{Q}s}{s^2 + \frac{\omega_0}{Q}s + \omega_0^2}$$

### Circuito A



## 1. Cálculo de Impedancias y Función de Transferencia $H(s)$

### Impedancias del circuito:
- Resistencia: $Z_R = R$
- Inductancia: $Z_L = sL$
- Capacitancia: $Z_C = \frac{1}{sC}$

Como la inductancia $L$ y la resistencia $R$ están conectadas en serie en la primera rama:
$$Z_1 = Z_L + Z_R = sL + R$$

La tensión de salida $V_2(s)$ se mide en paralelo sobre el capacitor:
$$Z_2 = Z_C = \frac{1}{sC}$$

### Aplicando el divisor de impedancias:
$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_2}{Z_1 + Z_2}$$

$$H(s) = \frac{\frac{1}{sC}}{sL + R + \frac{1}{sC}}$$

Multiplicando numerador y denominador por $sC$:
$$H(s) = \frac{1}{s^2 LC + sCR + 1}$$

Dividiendo numerador y denominador por $LC$ para llevar a la forma canónica:
$$H(s) = \frac{\frac{1}{LC}}{s^2 + s\frac{R}{L} + \frac{1}{LC}}$$

---

## 2. Determinación de Parámetros y Componentes ($L$ y $C$)

Comparando con la forma canónica de un filtro de segundo orden:
$$H(s) = \frac{\omega_0^2}{s^2 + \frac{\omega_0}{Q}s + \omega_0^2}$$

Identificamos las relaciones:
$$\omega_0^2 = \frac{1}{LC}$$
$$\frac{\omega_0}{Q} = \frac{R}{L}$$

A partir de los datos dados ($\omega_0 = 1 \text{ rad/s}$, $R = 1\ \Omega$, $\frac{\omega_0}{Q} = \sqrt{2}$):

1. **Cálculo de $L$:**
   $$\frac{\omega_0}{Q} = \frac{R}{L} \implies \sqrt{2} = \frac{1}{L} \implies \boxed{L = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2} \text{ H}}$$

2. **Cálculo de $C$:**
   $$\omega_0^2 = \frac{1}{LC} \implies 1^2 = \frac{1}{\left(\frac{\sqrt{2}}{2}\right) C} \implies \boxed{C = \frac{2}{\sqrt{2}} = \sqrt{2} \text{ F}}$$

Sustituyendo estos valores en la función de transferencia:
$$\boxed{H(s) = \frac{1}{s^2 + \sqrt{2}s + 1}}$$

---

## 3. Respuesta en Frecuencia ($s = j\omega$)

Evaluamos $H(s)$ reemplazando $s = j\omega$:
$$H(j\omega) = \frac{1}{(j\omega)^2 + \sqrt{2}(j\omega) + 1} = \frac{1}{1 - \omega^2 + j\sqrt{2}\omega}$$

---

### A) Cálculo del Módulo $|H(j\omega)|$

$$|H(j\omega)| = \frac{|1|}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{1}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}}$$

Desarrollando el binomio del denominador:
$$(1 - \omega^2)^2 + 2\omega^2 = 1 - 2\omega^2 + \omega^4 + 2\omega^2 = 1 + \omega^4$$

Por lo tanto:
$$\boxed{|H(j\omega)| = \frac{1}{\sqrt{1 + \omega^4}}}$$

#### Evaluación del módulo y respuesta asintótica en decibeles ($G_{\text{dB}}$):

$$ G_{\text{dB}}(\omega) = 20 \log_{10}\left(|H(j\omega)|\right) $$

1. **Para $\omega = 1$ rad/s:**
   $$|H(j1)| = \frac{1}{\sqrt{1^4 + 1}} = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2}$$
   $$G_{\text{dB}}(1) = 20 \log_{10}\left(\frac{\sqrt{2}}{2}\right) \approx -3.01 \text{ dB}$$

2. **Bajas frecuencias ($\omega \ll 1$):**
   $$\lim_{\omega \to 0} |H(j\omega)| \approx 1$$
   $$G_{\text{dB}}(\omega) = 20 \log_{10}(1) = 0 \text{ dB}$$

3. **Altas frecuencias ($\omega \gg 1$):**
   $$|H(j\omega)| \approx \frac{1}{\sqrt{\omega^4}} = \frac{1}{\omega^2}$$
   $$G_{\text{dB}}(\omega) \approx 20 \log_{10}\left(\frac{1}{\omega^2}\right) = -40 \log_{10}(\omega) \text{ dB}$$
   *(Pendiente asintótica de $-40 \text{ dB/década}$)*

---

### B) Cálculo de la Fase $\varphi(\omega)$

Utilizando la propiedad $\arg\left(\frac{N}{D}\right) = \arg(N) - \arg(D)$:
$$\varphi(\omega) = \arg(1) - \arg\left((1 - \omega^2) + j\sqrt{2}\omega\right)$$
$$\boxed{\varphi(\omega) = -\arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)}$$

#### Límites característicos de fase:
- **Para $\omega \to 0$:**
  $$\lim_{\omega \to 0} \varphi(\omega) = -\arctan(0) = 0^\circ$$

- **Para $\omega = 1$:**
  $$\varphi(1) = -\arg(0 + j\sqrt{2}) = -90^\circ$$

- **Para $\omega \to \infty$:**
  $$\lim_{\omega \to \infty} \varphi(\omega) = -180^\circ$$

---

## 4. Clasificación del Filtro

Analizamos la respuesta en frecuencia en los extremos del espectro:
- $\lim_{\omega \to 0} |H(j\omega)| = 1$ (deja pasar la continua y bajas frecuencias).
- $\lim_{\omega \to \infty} |H(j\omega)| = 0$ (atenuación total a altas frecuencias).

Por lo tanto, se trata de un **FILTRO PASA-BAJOS DE SEGUNDO ORDEN**.

---

## 5. Frecuencia de Corte ($\omega_c$) y Ancho de Banda ($\text{BW}$)

### Frecuencia de Corte ($\omega_c$):
Es la frecuencia donde la potencia cae a la mitad, es decir, donde $|H(j\omega_c)| = \frac{1}{\sqrt{2}}$:

$$\frac{1}{\sqrt{1 + \omega_c^4}} = \frac{1}{\sqrt{2}}$$

Elevando al cuadrado ambos miembros:
$$1 + \omega_c^4 = 2 \implies \omega_c^4 = 1 \implies \boxed{\omega_c = 1 \text{ rad/s}}$$

### Ancho de Banda ($\text{BW}$):
Para un filtro pasa-bajos, la banda pasante abarca desde $\omega = 0$ hasta $\omega = \omega_c$:

$$\text{BW} = \omega_c - 0 = 1 - 0 \implies \boxed{\text{BW} = 1 \text{ rad/s}}$$

### Circuito B


## 1. Cálculo de Impedancias y Función de Transferencia $H(s)$

### Impedancias del circuito:
- Rama de entrada ($C$ y $R$ en paralelo):
  $$Z_1 = \frac{R \cdot \frac{1}{sC}}{\frac{1}{sC} + R} = \frac{R}{1 + sRC}$$

- Rama de salida ($L$ en paralelo a masa):
  $$Z_2 = Z_L = sL$$

### Divisor de tensión:
$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_2}{Z_1 + Z_2} = \frac{sL}{\frac{R}{1 + sRC} + sL}$$

Multiplicando el denominador por $(1 + sRC)$:
$$H(s) = \frac{sL}{\frac{R + sL(1 + sRC)}{1 + sRC}} = \frac{sL(1 + sRC)}{R + sL + s^2 LRC} = \frac{s^2 LRC + sL}{s^2 LRC + sL + R}$$

Dividiendo numerador y denominador por $LRC$ para obtener la forma canónica:
$$H(s) = \frac{s^2 + s \cdot \frac{1}{RC}}{s^2 + s \cdot \frac{1}{RC} + \frac{1}{LC}}$$

---

## 2. Determinación de Parámetros y Componentes ($L$ y $C$)

De la comparación con la forma canónica:
$$\omega_0^2 = \frac{1}{LC} \implies LC = 1$$
$$\frac{\omega_0}{Q} = \frac{1}{RC}$$

Con los valores dados ($R = 1\ \Omega$, $\omega_0 = 1 \text{ rad/s}$ y $\frac{\omega_0}{Q} = \sqrt{2}$):

1. **Cálculo de $C$:**
   $$\frac{1}{RC} = \sqrt{2} \implies \frac{1}{1 \cdot C} = \sqrt{2} \implies \boxed{C = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2} \text{ F}}$$

2. **Cálculo de $L$:**
   $$LC = 1 \implies L \cdot \frac{\sqrt{2}}{2} = 1 \implies \boxed{L = \frac{2}{\sqrt{2}} = \sqrt{2} \text{ H}}$$

Reemplazando en la función de transferencia:
$$\boxed{H(s) = \frac{s^2 + \sqrt{2}s}{s^2 + \sqrt{2}s + 1}}$$

## 3. Respuesta en Frecuencia ($s = j\omega$)

Sustituyendo $s = j\omega$:
$$H(j\omega) = \frac{(j\omega)^2 + \sqrt{2}(j\omega)}{(j\omega)^2 + \sqrt{2}(j\omega) + 1} = \frac{-\omega^2 + j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$

---

### A) Cálculo del Módulo $|H(j\omega)|$

$$|H(j\omega)| = \frac{\sqrt{(-\omega^2)^2 + (\sqrt{2}\omega)^2}}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}} = \frac{\sqrt{\omega^4 + 2\omega^2}}{\sqrt{1 - 2\omega^2 + \omega^4 + 2\omega^2}}$$

$$\boxed{|H(j\omega)| = \frac{\omega\sqrt{\omega^2 + 2}}{\sqrt{1 + \omega^4}} \quad (\text{para } \omega \ge 0)}$$

#### Evaluación del módulo y respuesta asintótica en decibeles ($G_{\text{dB}}$):

1. **En $\omega = 1$ rad/s:**
   $$|H(j1)| = \frac{1 \cdot \sqrt{1^2 + 2}}{\sqrt{1 + 1^4}} = \frac{\sqrt{3}}{\sqrt{2}} = \sqrt{\frac{3}{2}}$$
   $$G_{\text{dB}}(1) = 20 \log_{10}\left(\sqrt{\frac{3}{2}}\right) \approx 1.76 \text{ dB}$$

2. **Bajas frecuencias ($\omega \ll 1$):**
   $$\lim_{\omega \to 0} |H(j\omega)| = 0$$
   $$|H(j\omega)| \approx \frac{\omega\sqrt{2}}{1} = \sqrt{2}\omega$$
   $$G_{\text{dB}}(\omega) \approx 20 \log_{10}(\sqrt{2}\omega) = 20 \log_{10}(\sqrt{2}) + 20 \log_{10}(\omega) \approx 3.01 + 20 \log_{10}(\omega) \text{ dB}$$
   *(Pendiente asintótica de $+20 \text{ dB/década}$)*

3. **Altas frecuencias ($\omega \gg 1$):**
   $$\lim_{\omega \to \infty} |H(j\omega)| = 1$$
   $$|H(j\omega)| \approx \frac{\omega \cdot \omega}{\omega^2} = 1$$
   $$G_{\text{dB}}(\omega) \approx 20 \log_{10}(1) = 0 \text{ dB}$$

---

### B) Cálculo de la Fase $\varphi(\omega)$

Utilizando la propiedad $\arg\left(\frac{A}{B}\right) = \arg(A) - \arg(B)$:
$$\varphi(\omega) = \arg(-\omega^2 + j\sqrt{2}\omega) - \arg(1 - \omega^2 + j\sqrt{2}\omega)$$

#### Evaluación puntual y límites:
- **Para $\omega = 1$:**
  $$\varphi(1) = \arg(-1 + j\sqrt{2}) - \arg(0 + j\sqrt{2})$$
  $$\varphi(1) = \left[\pi - \arctan(\sqrt{2})\right] - \frac{\pi}{2} = \frac{\pi}{2} - \arctan(\sqrt{2}) \approx 90^\circ - 54.74^\circ \approx 35.26^\circ$$

- **Bajas frecuencias ($\omega \to 0$):**
  $$\varphi(\omega) = \pi - \arctan\left(\frac{\sqrt{2}}{\omega}\right) - \arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)$$
  $$\lim_{\omega \to 0} \varphi(\omega) = \pi - \frac{\pi}{2} - 0 = \frac{\pi}{2} = 90^\circ$$

- **Altas frecuencias ($\omega \to \infty$):**
  $$\lim_{\omega \to \infty} \varphi(\omega) = \pi - 0 - \pi = 0^\circ$$

## 4. Clasificación del Filtro

Analizando los límites para frecuencias extremas:
- $\lim_{\omega \to 0} |H(j\omega)| = 0$ (bloquea la componente continua y bajas frecuencias).
- $\lim_{\omega \to \infty} |H(j\omega)| = 1$ (deja pasar las altas frecuencias).

Por lo tanto, se trata de un **FILTRO PASA-ALTOS DE SEGUNDO ORDEN**.

## 5. Frecuencia de Corte ($\omega_c$) y Ancho de Banda ($\text{BW}$)

### Frecuencia de Corte ($\omega_c$):
Igualamos el módulo a $\frac{1}{\sqrt{2}}$:
$$\frac{\omega_c\sqrt{\omega_c^2 + 2}}{\sqrt{1 + \omega_c^4}} = \frac{1}{\sqrt{2}}$$

Elevando ambos lados al cuadrado:
$$\frac{\omega_c^2 (\omega_c^2 + 2)}{1 + \omega_c^4} = \frac{1}{2} \implies \frac{\omega_c^4 + 2\omega_c^2}{1 + \omega_c^4} = \frac{1}{2}$$

Multiplicando cruzado:
$$2(\omega_c^4 + 2\omega_c^2) = 1 + \omega_c^4 \implies \omega_c^4 + 4\omega_c^2 - 1 = 0$$

Haciendo la sustitución $u = \omega_c^2 > 0$:
$$u^2 + 4u - 1 = 0$$

Aplicando la fórmula cuadrática resolvente:
$$u = \frac{-4 + \sqrt{4^2 - 4(1)(-1)}}{2} = \frac{-4 + \sqrt{20}}{2} = -2 + \sqrt{5} \approx 0.236$$
*(Descartamos la raíz negativa por ser un absurdo).*

Despejando $\omega_c$:
$$\omega_c = \sqrt{-2 + \sqrt{5}} \approx \sqrt{0.236} \implies \boxed{\omega_c \approx 0.48 \text{ rad/s}}$$

### Ancho de Banda ($\text{BW}$):
En un filtro pasa-altos ideal, el ancho de banda abarca desde la frecuencia de corte hasta infinito:
$$\text{BW} = \omega_{\text{superior}} - \omega_c \to \infty$$

### Circuito C


## 1. Cálculo de Impedancias y Función de Transferencia $H(s)$

### Impedancias del circuito:
- Rama en serie de entrada:
  $$Z_1 = R$$

- Rama en paralelo a masa ($L$ y $C$ en paralelo):
  $$Z_2 = Z_L \parallel Z_C = \frac{sL \cdot \frac{1}{sC}}{sL + \frac{1}{sC}} = \frac{\frac{sL}{sC}}{sL + \frac{1}{sC}} \cdot \frac{sC}{sC} = \frac{sL}{1 + s^2 LC}$$

### Divisor de tensión:
$$H(s) = \frac{V_2(s)}{V_1(s)} = \frac{Z_2}{Z_1 + Z_2} = \frac{\frac{sL}{1 + s^2 LC}}{R + \frac{sL}{1 + s^2 LC}}$$

Multiplicando numerador y denominador por $(1 + s^2 LC)$:
$$H(s) = \frac{sL}{R(1 + s^2 LC) + sL} = \frac{sL}{s^2 LCR + sL + R}$$

Dividiendo numerador y denominador por $LCR$ para llevar a la forma canónica:
$$H(s) = \frac{\frac{sL}{LCR}}{\frac{s^2 LCR}{LCR} + \frac{sL}{LCR} + \frac{R}{LCR}} = \frac{s \cdot \frac{1}{CR}}{s^2 + s \cdot \frac{1}{CR} + \frac{1}{LC}}$$

---

## 2. Determinación de Parámetros y Componentes ($L$ y $C$)

Comparando con la forma canónica de un filtro pasa-banda de segundo orden:
$$H(s) = \frac{\frac{\omega_0}{Q}s}{s^2 + \frac{\omega_0}{Q}s + \omega_0^2}$$

Identificamos las relaciones:
$$\omega_0^2 = \frac{1}{LC} \implies LC = 1$$
$$\frac{\omega_0}{Q} = \frac{1}{CR}$$

Con los valores dados ($\omega_0 = 1 \text{ rad/s}$, $R = 1\ \Omega$ y $\frac{\omega_0}{Q} = \sqrt{2}$):

1. **Cálculo de $C$:**
   $$\frac{1}{CR} = \sqrt{2} \implies \frac{1}{C \cdot 1} = \sqrt{2} \implies \boxed{C = \frac{1}{\sqrt{2}} = \frac{\sqrt{2}}{2} \text{ F}}$$

2. **Cálculo de $L$:**
   $$LC = 1 \implies L \cdot \frac{\sqrt{2}}{2} = 1 \implies \boxed{L = \frac{2}{\sqrt{2}} = \sqrt{2} \text{ H}}$$

Sustituyendo estos valores en la función de transferencia:
$$\boxed{H(s) = \frac{\sqrt{2}s}{s^2 + \sqrt{2}s + 1}}$$

## 3. Respuesta en Frecuencia ($s = j\omega$)

Sustituyendo $s = j\omega$:
$$H(j\omega) = \frac{\sqrt{2}(j\omega)}{(j\omega)^2 + \sqrt{2}(j\omega) + 1} = \frac{j\sqrt{2}\omega}{(1 - \omega^2) + j\sqrt{2}\omega}$$

---

### A) Cálculo del Módulo $|H(j\omega)|$

$$|H(j\omega)| = \frac{|\sqrt{2}\omega j|}{|(1 - \omega^2) + j\sqrt{2}\omega|} = \frac{\sqrt{2}\omega}{\sqrt{(1 - \omega^2)^2 + (\sqrt{2}\omega)^2}}$$

Desarrollando el denominador:
$$(1 - \omega^2)^2 + 2\omega^2 = 1 - 2\omega^2 + \omega^4 + 2\omega^2 = 1 + \omega^4$$

Por lo tanto:
$$\boxed{|H(j\omega)| = \frac{\sqrt{2}\omega}{\sqrt{1 + \omega^4}}}$$

#### Evaluación del módulo y respuesta asintótica en decibeles ($G_{\text{dB}}$):

1. **Frecuencia central de resonancia ($\omega = 1$ rad/s):**
   $$|H(j1)| = \frac{\sqrt{2}(1)}{\sqrt{1 + 1^4}} = \frac{\sqrt{2}}{\sqrt{2}} = 1$$
   $$G_{\text{dB}}(1) = 20 \log_{10}(1) = 0 \text{ dB}$$

2. **Bajas frecuencias ($\omega \ll 1$):**
   $$\lim_{\omega \to 0} |H(j\omega)| = 0$$
   $$|H(j\omega)| \approx \sqrt{2}\omega$$
   $$G_{\text{dB}}(\omega) \approx 20 \log_{10}(\sqrt{2}\omega) = 20 \log_{10}(\sqrt{2}) + 20 \log_{10}(\omega) \approx 3.01 + 20 \log_{10}(\omega) \text{ dB}$$
   *(Pendiente asintótica de $+20 \text{ dB/década}$)*

3. **Altas frecuencias ($\omega \gg 1$):**
   $$\lim_{\omega \to \infty} |H(j\omega)| = 0$$
   $$|H(j\omega)| \approx \frac{\sqrt{2}\omega}{\sqrt{\omega^4}} = \frac{\sqrt{2}}{\omega}$$
   $$G_{\text{dB}}(\omega) \approx 20 \log_{10}\left(\frac{\sqrt{2}}{\omega}\right) = 20 \log_{10}(\sqrt{2}) - 20 \log_{10}(\omega) \approx 3.01 - 20 \log_{10}(\omega) \text{ dB}$$
   *(Pendiente asintótica de $-20 \text{ dB/década}$)*

---

### B) Cálculo de la Fase $\varphi(\omega)$

Utilizando la propiedad $\arg\left(\frac{N}{D}\right) = \arg(N) - \arg(D)$:
$$\varphi(\omega) = \arg(j\sqrt{2}\omega) - \arg\left((1 - \omega^2) + j\sqrt{2}\omega\right)$$
$$\boxed{\varphi(\omega) = 90^\circ - \arctan\left(\frac{\sqrt{2}\omega}{1 - \omega^2}\right)}$$

#### Evaluación puntual y límites:
- **Bajas frecuencias ($\omega \to 0$):**
  $$\lim_{\omega \to 0} \varphi(\omega) = 90^\circ - \arctan(0) = 90^\circ$$

- **En resonancia ($\omega = 1$):**
  $$\varphi(1) = 90^\circ - \arctan\left(\frac{\sqrt{2}}{0}\right) = 90^\circ - 90^\circ = 0^\circ$$

- **Altas frecuencias ($\omega \to \infty$):**
  $$\lim_{\omega \to \infty} \varphi(\omega) = 90^\circ - 180^\circ = -90^\circ$$

## 4. Clasificación del Filtro

Analizando la respuesta en frecuencia:
- $\lim_{\omega \to 0} |H(j\omega)| = 0$
- $\lim_{\omega \to \infty} |H(j\omega)| = 0$
- Presenta un valor máximo de ganancia en la frecuencia central $\omega_0 = 1 \text{ rad/s}$ ($|H(j1)| = 1$).

Por lo tanto, se trata de un **FILTRO PASA-BANDA DE SEGUNDO ORDEN**.

## 5. Frecuencias de Corte ($\omega_{c1}, \omega_{c2}$) y Ancho de Banda ($\text{BW}$)

### Frecuencias de Corte ($\omega_c$):
Igualamos el módulo a $\frac{1}{\sqrt{2}}$:
$$\frac{\sqrt{2}\omega_c}{\sqrt{1 + \omega_c^4}} = \frac{1}{\sqrt{2}}$$

Elevando ambos miembros al cuadrado:
$$\frac{2\omega_c^2}{1 + \omega_c^4} = \frac{1}{2} \implies 4\omega_c^2 = 1 + \omega_c^4 \implies \omega_c^4 - 4\omega_c^2 + 1 = 0$$

Haciendo el cambio de variable $u = \omega_c^2 > 0$:
$$u^2 - 4u + 1 = 0$$

Aplicando la fórmula cuadrática resolvente:
$$u = \frac{4 \pm \sqrt{(-4)^2 - 4(1)(1)}}{2} = \frac{4 \pm \sqrt{12}}{2} = 2 \pm \sqrt{3}$$

Despejando ambas raíces reales positivas:

1. **Frecuencia de corte inferior ($\omega_{c1}$):**
   $$\omega_{c1} = \sqrt{2 - \sqrt{3}} = \sqrt{2 - 1.732} = \sqrt{0.268} \implies \boxed{\omega_{c1} \approx 0.51 \text{ rad/s}}$$

2. **Frecuencia de corte superior ($\omega_{c2}$):**
   $$\omega_{c2} = \sqrt{2 + \sqrt{3}} = \sqrt{2 + 1.732} = \sqrt{3.732} \implies \boxed{\omega_{c2} \approx 1.93 \text{ rad/s}}$$

### Ancho de Banda ($\text{BW}$):
El ancho de banda se define como la diferencia entre la frecuencia de corte superior e inferior:
$$\text{BW} = \omega_{c2} - \omega_{c1} = 1.93 - 0.51 \implies \boxed{\text{BW} = 1.42 \text{ rad/s}}$$
